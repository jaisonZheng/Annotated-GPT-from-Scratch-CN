```python
def get_statics(idx):
    counts = {}
    # 在UTF-8字节流当中，得到所有相邻对的出现频率
    for pair in zip(idx, idx[1:]): # 用zip得到相邻pair，非常pythonic的写法。
        counts[pair] = counts.get(pair, 0) + 1
    return counts

print (sorted(((val, key) for key, val in get_statics(idx).items()), reverse = True))
# 将val(频次)放到元组第一个（python默认以第一个元素来进行排序）
```

```python
def replace(indexs, pair, newIndex):
    newIndexs = []
    i = 0
    while i < len(indexs):
        if (i + 1) < len(indexs) and (indexs[i], indexs[i + 1]) == pair:
            newIndexs.append(newIndex)
            i += 2
        else:
            newIndexs.append(indexs[i])
            i += 1
    return newIndexs
```

```python
vocab_size = 276
iter_times = vocab_size - 256

merges = {} #(int, int) -> int
for i in range(iter_times):
    stats = get_statics(idx)
    maxPair = max(stats, key = stats.get) 
    # 第一个参数（必须是可迭代对象）决定了返回的结果，第二个参数决定了max的判断条件
    newIndex = 256 + i
    merges[maxPair] = newIndex
    idx = replace(idx, maxPair, newIndex)
    print(f"Merge {maxPair} into {newIndex}")
```

```python
vocab = {index:bytes([index]) for index in range(256)}
# 0 - 255的字节编码是固定的，而 > 255的bytes是由这255位拼接而成的
for (p0, p1), idx in merges.items():
    vocab[idx] = vocab[p0] + vocab[p1] # 这里是拼接而不是int加和

def decode(indexs):
    # 给定字节序列indexs，返回str。
    tokens = b"".join(vocab[idx] for idx in indexs)
    text = tokens.decode("utf-8", errors = 'replace') 
    # 这个error处理体现了LLM对token的decode处理的特性。
    # 即可能会出现无法decode的（不符合utf-8格式）
    # 而decode函数本身默认的error处理是"strict"，就会直接抛出错误
    # 但这种错误是无法避免的，所以我们使用replace来替换掉该字符（变成错误符号）
    return text
```

```python
def encode(str):
    indexs = list(str.encode("utf-8"))
    while len(indexs) >= 2:
        stats = get_statics(indexs)
        pair = min(stats, key=lambda p: merges.get(p, float('inf'))) 
        # 以merges中的rank为指标来取stats中的“最小值”，如果stats当前所有的值都不在merges的pair对里，返回stats的第一个值（python默认行为）
        if pair not in merges:
            break
        index = merges[pair]
        indexs = replace(indexs, pair, index)
    return indexs


print(encode("hello world，， 。"))
```
